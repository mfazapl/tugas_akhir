```python
import os
import shutil
```

```python
data_path=r"D:\datas_suara_split\Real\tes"
output_folder=r"D:\datas_suara_split\Real\sementara"
```

```python

os.makedirs(output_folder, exist_ok=True)

for name in os.listdir(data_path):
    src_path= os.path.join(data_path, name)

    if not os.path.isfile(src_path):
        continue

    filename, ext= os.path.splitext(name)
    newfile= f"{filename}_test{ext}"

    new_path= os.path.join(output_folder, newfile)

    shutil.move(src_path, new_path)
```
