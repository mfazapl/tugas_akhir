```python
import os
import shutil
```

```python
dir_name= r"C:\Users\Hype G12\OneDrive\Desktop\tts_wikipedia"
dir_name_output= r"C:\Users\Hype G12\OneDrive\Desktop\tts_wikiped"
```

```python
for count, data in enumerate(os.listdir(dir_name)):
   src = os.path.join(dir_name, data)
   ext= os.path.splitext(data)[1]
   new_name= f"Sintesis_{count+1:03d}{ext}"
   output= os.path.join(dir_name_output, new_name)

   shutil.move(src, output)


```
