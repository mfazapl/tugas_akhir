```python
import random
import os
import shutil
random.seed(48)
```

```python
dt_path=r"D:\DATA_SUARA\Real"
dt_path_dist=r"D:\datas_suara_split\Real"
splits={
    "train": 0.70,
    "val": 0.15,
    "test": 0.15
}

aduio_ext= ".wav"


files= [
    f for f in os.listdir(dt_path)
    if f.lower().endswith(aduio_ext)
]

if not files:
    RuntimeError("Tidak ada file Wav")

random.shuffle(files)


total= len(files)
n_val= int(total *splits["val"])
n_train= int(total* splits["train"])

split_files={
    "train": files[:n_train],
    "val": files[n_train: n_train+n_val],
    "test": files[n_train+n_val:]
}

for split, file_list in split_files.items():
    dst_dir = os.path.join(dt_path_dist, split)
    os.makedirs(dst_dir, exist_ok=True)

    for f in file_list:
        shutil.copy(
        os.path.join(dt_path, f),
        os.path.join(dst_dir, f)
    )
```
