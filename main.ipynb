```python
import random
import shutil
import os

```

```python
dt_path=r"D:\data_extract"
dt_path_dist=r"D:\training\skenario\kelima"
splits={
    "train": 0.90,
    "val": 0.05,
    "test": 0.05
}

img_ext= ".png"


for cls in os.listdir(dt_path):
    cls_path= os.path.join(dt_path, cls)

    if not os.path.isdir(cls_path):
        continue

    files= [
        f for f in os.listdir(cls_path)
        if f.lower().endswith(img_ext)
    ]
```

```python
random.seed(48)
random.shuffle(files)

total= len(files)
n_val= int(total *splits["val"])
n_train= int(total* splits["train"])

split_files={
    "train": files[:n_train],
    "val": files[n_train: n_train+n_val],
    "test": files[n_train+n_val:]
}

for split, file_list in split_files.items():
    dst_dir = os.path.join(dt_path_dist, split, cls)
    os.makedirs(dst_dir, exist_ok=True)

    for f in file_list:
        shutil.copy(
            os.path.join(cls_path, f),
            os.path.join(dst_dir, f)
        )
print(f"{cls}: {total}")
```
